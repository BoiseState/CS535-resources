```python
from pyspark.sql import SparkSession
```

```python
from pyspark.sql.functions import avg, col
```

```python
spark = SparkSession.builder.master('local[*]').appName("sample").getOrCreate()
```

```python
sc = spark.sparkContext
```

```python
dataRDD = sc.parallelize([("Brooke", 20),("Denny", 31), ("Jules", 30), ("TD", 35), ("Brooke", 45)])
```

```python
agesRDD = (dataRDD
           .map(lambda x: (x[0], (x[1], 1)))
           .reduceByKey(lambda x, y: (x[0] + y[0], x[1]+y[1]))
           .map(lambda x: (x[0] , x[1][0]/x[1][1])))
```

```python
result = agesRDD.collect()
```

```python
for val in result:
    print(val)
```

```python
spark = (SparkSession
         .builder
         .appName("AuthorAges")
         .getOrCreate())
```

```python
data_df = spark.createDataFrame([("Brooke", 20), ("Denny", 31), ("Jules", 30), ("TD", 35), ("Brooke", 45)], ["name", "age"])
```

```python
avg_df = data_df.groupBy("name").agg(avg("age"))
```
